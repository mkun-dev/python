```python
#函数
def calculate_BMI(weight, height):
    BMI = float(weight) / float(height ** 2)
    if BMI <= 18.5:
        print('您偏瘦') 
    elif BMI <= 25:
        print('您很正常')
    elif BMI <= 30:
        print('您偏胖')
    else:
        print('真是头大肥猪')
    return BMI
calculate_BMI(67, 1.76)
```

```python
#定义cat类
class CuteCat:
    def __init__(self, cat_name, cat_age, cat_color):
        self.name = cat_name
        self.age = cat_age
        self.color = cat_color
        
    def speak(self):
        print("喵"*self.age)

cat1 = CuteCat('mimi',3 ,'white')

#定义学生类
class Student:
    def __init__(self, student_name, student_id):
        self.name  = student_name
        self.id = student_id
        self.grades = {'语文':0 ,'数学':0 , '英语':0}
    
    def set_grade(self, course, grade):
        if course in self.grades:
            self.grades[course] = grade
        
    def print_grades(self):
        print(f"学生{self.name} (学号：{self.id})的成绩为：")
        for course in self.grades:
            print(f"{course}:{self.grades[course]}分")
        
student1 = Student('mk','123') 
student1.set_grade('数学', 120)
student1.print_grades()
```

```python
class Employee:
    def __init__(self, employee_name, employee_id):
        self.name = employee_name
        self.id = employee_id
        
    def print_info(self):
        print(f"员工的名字为{self.name}\n员工的工号为{self.id}")

class FullTimeEmployee(Employee):
    def __init__(self, employee_name, employee_id, mounthly_salary):
        super().__init__(employee_name, employee_id)
        self.salary = mounthly_salary
    def calculate_mounthly_salary(self):
        return self.salary
    
class PartTimeEmployee(Employee):
    def __init__(self, name, id, daily_salary, work_days):
        super().__init__(name, id)
        self.salary = daily_salary
        self.days = work_days
    
    def calculate_mounthly_salary(self):
        return self.salary * self.days
    
zhangsan = FullTimeEmployee('张三', '1001' , 6000)
lisi = PartTimeEmployee('李四', '1002',100 , 30)
zhangsan.print_info()
print(zhangsan)
```
